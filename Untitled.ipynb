```python
import pandas as pd
data = {
    "employee_id": ["E101","E102","E103","E104","E105",
                    "E106","E107","E108","E109","E110",
                    "E111","E112","E113","E114","E115"],
    "name": ["Amit","Neha","Rahul","Priya","Karan",
             "Simran","Vikas","Anjali","Rohit","Sneha",
             "Arjun","Pooja","Manish","Kavita","Nitin"],
    "department": ["IT","HR","IT","Finance","HR",
                    "IT","Finance","IT","HR","Finance",
                    "IT","HR","Finance","IT","Finance"],

    "age": [25,29,32,35,27,30,38,26,34,31,40,28,36,33,42],

    "experience_years": [1,4,7,10,3,6,13,2,9,5,15,3,11,8,17],

    "salary": [42000,55000,78000,95000,50000,
               72000,125000,46000,88000,68000,
               140000,52000,110000,90000,155000],

    "performance_score": [65,72,85,88,70,
                          82,95,68,86,78,
                          97,74,91,89,96],

    "year": [2022,2022,2022,2022,2023,
             2023,2023,2023,2024,2024,
             2024,2024,2025,2025,2025]
}

df = pd.DataFrame(data)
df

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sns.set_theme(style="whitegrid")

# 1. Histogram with KDE
plt.figure(figsize=(7,4))

sns.histplot(df["salary"], kde=True)

plt.title("Distribution of Employee Salary")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()


# 2. Boxplot by Department
plt.figure(figsize=(7,4))

sns.boxplot(
    x="department",
    y="salary",
    data=df
)

plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary")

plt.show()


# 3. Scatter Plot
plt.figure(figsize=(7,4))

sns.scatterplot(
    x="experience_years",
    y="salary",
    hue="department",
    data=df,
    s=100
)

plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")

plt.show()


# 4. Correlation Heatmap
corr = df.corr(numeric_only=True)

plt.figure(figsize=(9,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()


# 5. Strongest correlation
corr_pairs = corr.where(
    ~pd.DataFrame(
        __import__("numpy").tril(
            __import__("numpy").ones(corr.shape),
            k=-1
        ),
        index=corr.index,
        columns=corr.columns
    ).astype(bool)
)

corr_pairs = corr_pairs.stack().sort_values(
    key=abs,
    ascending=False
)

print("Strongest correlation pair:")
print(corr_pairs.head(1))

```

```python
import plotly.express as px

# ----------------------------------
# 1. Interactive Scatter Plot
# ----------------------------------

fig1 = px.scatter(
    df,
    x="experience_years",
    y="salary",
    color="department",
    hover_data=[
        "employee_id",
        "name",
        "age",
        "performance_score"
    ],
    title="Interactive Experience vs Salary"
)
fig1.update_layout(
    xaxis_title="Experience (Years)",
    yaxis_title="Salary",
    hovermode="closest"
)

fig1.show()


# ----------------------------------
# 2. Interactive Bar Chart
# ----------------------------------

department_salary = (
    df.groupby("department", as_index=False)["salary"]
      .mean()
)

fig2 = px.bar(
    department_salary,
    x="department",
    y="salary",
    color="department",
    text_auto=".2s",
    title="Average Salary by Department"
)

fig2.update_layout(
    xaxis_title="Department",
    yaxis_title="Average Salary"
)

fig2.show()


# ----------------------------------
# 3. Interactive Line Chart
# ----------------------------------

year_performance = (
    df.groupby("year", as_index=False)["performance_score"]
      .mean()
)

fig3 = px.line(
    year_performance,
    x="year",
    y="performance_score",
    markers=True,
    title="Average Performance Score by Year"
)

fig3.update_layout(
    xaxis_title="Year",
    yaxis_title="Average Performance Score"
)

fig3.show()


# ----------------------------------
# 4. Export HTML files
# ----------------------------------

fig1.write_html("interactive_scatter.html")
fig2.write_html("department_salary.html")
fig3.write_html("performance_trend.html")

print("All interactive charts exported successfully.")
```
