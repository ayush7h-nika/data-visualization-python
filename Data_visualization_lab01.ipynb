```python
import io
import pandas as pd

# ==========================================================
# STEP 1: Create/Read CSV dataset into csv_df
# ==========================================================

# Simulated CSV data stream
csv_data = """id,name,age,salary,email
101,Amit,20,45000.0,amit@gmail.com
102,Priya,21,55000.0,priya@gmail.com
103,Rahul,19,40000.0,rahul@gmail.com
104,Neha,22,60000.0,neha@gmail.com
105,Vikas,20,50000.0,vikas@gmail.com
"""

# If loading an actual file on disk, replace the line below with:
# csv_df = pd.read_csv("your_file.csv")
csv_df = pd.read_csv(io.StringIO(csv_data))

# Make sure data types match the schema
csv_df["age"] = csv_df["age"].astype("int64")
csv_df["salary"] = csv_df["salary"].astype("float64")
csv_df["email"] = csv_df["email"].astype("object")


# ==========================================================
# STEP 2: Define Schema
# ==========================================================

schema = {
    "age": {"dtype": "int64", "min": 0, "max": 120},
    "salary": {"dtype": "float64", "min": 0, "max": 10_000_000},
    "email": {"dtype": "object"},
}


# ==========================================================
# STEP 3: Validate Dataset
# ==========================================================

validation_report = []

# Check for missing required columns
missing_cols = set(schema.keys()) - set(csv_df.columns)

if missing_cols:
    validation_report.append(f"FAIL: Missing required columns {missing_cols}")


# Check each column against schema
for col, rules in schema.items():

    if col not in csv_df.columns:
        continue

    # Check data type
    actual_dtype = str(csv_df[col].dtype)
    expected_dtype = rules["dtype"]

    dtype_ok = actual_dtype == expected_dtype

    # Check range
    range_ok = True
    violations = 0

    if "min" in rules and "max" in rules:

        in_range = csv_df[col].between(rules["min"], rules["max"])

        violations = (~in_range).sum()
        range_ok = violations == 0

    # Final status
    status = "PASS" if dtype_ok and range_ok else "FAIL"

    validation_report.append(
        f"{status}: {col} | "
        f"dtype={actual_dtype} "
        f"(expected {expected_dtype}) | "
        f"range_violations={violations}"
    )


# ==========================================================
# STEP 4: Display Validation Report
# ==========================================================

print("=" * 70)
print("              DATASET VALIDATION REPORT")
print("=" * 70)

print("\n".join(validation_report))


# ==========================================================
# STEP 5: Overall Validation
# ==========================================================

overall = (
    "PASS"
    if all("FAIL" not in r for r in validation_report)
    else "FAIL"
)

print("\n" + "=" * 70)
print("Overall dataset validation:", overall)
print("=" * 70)


# ==========================================================
# STEP 6: Display Dataset
# ==========================================================

print("\nCleaned Dataset:")
print(csv_df)
```
